```python
import numpy as np

# -----------------------------
# Input Sequence
# -----------------------------
X = [1, 2, 3, 4]

# -----------------------------
# Initialize Weights
# -----------------------------
W_xh = np.array([[0.5]])   # Input → Hidden
W_hh = np.array([[0.8]])   # Hidden → Hidden
W_hy = np.array([[1.2]])   # Hidden → Output

b_h = np.array([[0.1]])
b_y = np.array([[0.2]])

# Initial Hidden State
h_prev = np.zeros((1,1))

print("Forward Propagation Started\n")

# -----------------------------
# Forward Pass
# -----------------------------
for t, x in enumerate(X):

    x = np.array([[x]])

    # Hidden State Calculation
    h = np.tanh(np.dot(W_xh, x) +
                np.dot(W_hh, h_prev) +
                b_h)

    # Output Calculation
    y = np.dot(W_hy, h) + b_y

    print(f"Time Step {t+1}")
    print("Input:", x[0][0])
    print("Hidden State:", h[0][0])
    print("Output:", y[0][0])
    print("-"*40)

    h_prev = h
```

```python
import numpy as np

np.random.seed(42)

# Training Data
inputs = np.array([1,2,3,4]).reshape(-1,1)
targets = np.array([2,3,4,5]).reshape(-1,1)

# Hyperparameters
input_size = 1
hidden_size = 5
output_size = 1
learning_rate = 0.01
epochs = 1000

# Weights
Wxh = np.random.randn(hidden_size, input_size)
Whh = np.random.randn(hidden_size, hidden_size)
Why = np.random.randn(output_size, hidden_size)

bh = np.zeros((hidden_size,1))
by = np.zeros((output_size,1))

for epoch in range(epochs):

    h_prev = np.zeros((hidden_size,1))

    loss = 0

    for x, target in zip(inputs, targets):

        x = x.reshape(1,1)

        h = np.tanh(Wxh @ x + Whh @ h_prev + bh)

        y = Why @ h + by

        loss += np.sum((target-y)**2)

        # Gradient
        dy = y - target

        dWhy = dy @ h.T
        dby = dy

        dh = Why.T @ dy
        dh_raw = (1-h*h) * dh

        dWxh = dh_raw @ x.T
        dWhh = dh_raw @ h_prev.T
        dbh = dh_raw

        # Update
        Why -= learning_rate*dWhy
        by -= learning_rate*dby

        Wxh -= learning_rate*dWxh
        Whh -= learning_rate*dWhh
        bh -= learning_rate*dbh

        h_prev = h

    if epoch % 100 == 0:
        print(f"Epoch {epoch}, Loss: {loss}")

print("Training Completed")
```
