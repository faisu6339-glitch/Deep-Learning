#1. Create a Single Artificial Neuron

Input:

x = [2, 3]

Task:

Calculate weighted sum
Add bias
Print output

```python
x=[2,3]
w=[0.4,0.6]

b=0.5

weighted_sum=(x[0]*w[0])+(x[1]*w[1])
output=weighted_sum+b

print("weighted_sum:",weighted_sum)
print("Output:",output)
```

#Problem 2: Implement Sigmoid Function

```python
import math

def sigmoid(x):
    return 1 / (1 + math.exp(-x))

# Test values
test_values = [-2, -1, 0, 1, 2]

for value in test_values:
    print(f"sigmoid({value}) = {sigmoid(value):.4f}")
```

```python
def relu(x):
    return max(0, x)

test_values = [-2, -1, 0, 1, 2]

for value in test_values:
    print(f"relu({value}) = {relu(value)}")
```

#6. Plot Activation Functions

Plot:

Sigmoid
ReLU
Tanh
Leaky ReLU

using Matplotlib.

```python
import numpy as np
import matplotlib.pyplot as plt

x=np.linspace(-10,10,100)

```

```python
def sigmoid(x):
  return 1/(1 + np.exp(-x))
```

```python
def relu(x):
    return np.maximum(0, x)
```

```python
def tanh(x):
  return np.tanh(x)
```

```python
def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, alpha * x)
```

```python
plt.figure(figsize=(10, 6))

plt.plot(x, sigmoid(x), label='Sigmoid')
plt.plot(x, relu(x), label='ReLU')
plt.plot(x, tanh(x), label='Tanh')
plt.plot(x, leaky_relu(x), label='Leaky ReLU')

plt.title('Activation Functions')
plt.xlabel('Input (x)')
plt.ylabel('Output')
plt.grid(True)
plt.legend()
plt.show()
```
