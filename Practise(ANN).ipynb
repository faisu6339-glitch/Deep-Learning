#1. Create a Single Artificial Neuron

Input:

x = [2, 3]
w = [0.4, 0.6]
b = 0.5

Task:

Calculate weighted sum
Add bias
Print output

```python
x=[2,3]
w=[0.4,0.6]

b=0.5

weighted_sum=(x[0]*w[0])+(x[1]*w[1])
output=weighted_sum+b

print("weighted_sum:",weighted_sum)
print("Output:",output)
```

```python
import numpy as np

x = np.array([2, 3])
w = np.array([0.4, 0.6])
b = 0.5

output = np.dot(x, w) + b

print("Neuron Output:", output)
```

#Problem 2: Implement Sigmoid Function

```python
import math

def sigmoid(x):
    return 1 / (1 + math.exp(-x))

# Test values
test_values = [-2, -1, 0, 1, 2]

for value in test_values:
    print(f"sigmoid({value}) = {sigmoid(value):.4f}")
```

```python
def relu(x):
    return max(0, x)

test_values = [-2, -1, 0, 1, 2]

for value in test_values:
    print(f"relu({value}) = {relu(value)}")
```
