#RNN for Sequence Prediction

#Goal

Train an RNN to learn

10 20 30 → 40
20 30 40 → 50
30 40 50 → 60
40 50 60 → 70
50 60 70 → 80
60 70 80 → 90
70 80 90 → 100

Later we'll test it using

80 90 100 → ?
Expected = 110

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense
```

```python
X=np.array([
    [10,20,30],
    [20,30,40],
    [30,40,50],
    [40,50,60],
    [50,60,70],
    [60,70,80],
    [70,80,90]
])
y=np.array([
    40,
    50,
    60,
    70,
    80,
    90,
    100
])
```

```python
print(X)
print(y)
print(X.shape)
print(y.shape)
```

#Normalize Data

```python
X = X / 100
y = y / 100
```

```python
print(X)
print(y)
```

#Reshape for RNN

#RNN expects data in

(samples,
 timesteps,
 features)

Currently

(7,3)

Convert into

(7,3,1)
X = X.reshape((X.shape[0], X.shape[1], 1))

print(X.shape)

Output

(8,3,1)

Meaning::

a)8 samples,

Each sample has::

b)3 time steps,

Each timestep has::

d)1 feature

```python
X=X.reshape((X.shape[0], X.shape[1], 1))
print(X.shape)
```

#Build RNN Model

```python
model=Sequential()
model.add(SimpleRNN(50,activation='tanh',input_shape=(3,1)))
model.add(Dense(1))
model.summary()
```

#Explanation

SimpleRNN(20)

20 neurons

activation='tanh'

input_shape=(3,1)

3 timesteps

1 feature

#Dense layer

Predicts one number

#Compile Model

```python
model.compile(
    optimizer='adam',
    loss='mse'
)
```

```python
model.fit(
    X,
    y,
    epochs=500,
    verbose=1
)
```

# Predict New Sequence

We know

80 90 100

Next should be

110

```python
test=np.array([80,90,100])
test=test/100
test=test.reshape((1,3,1))
```

```python
prediction = model.predict(test)

print("Predicted :", prediction[0][0]*100)
```
