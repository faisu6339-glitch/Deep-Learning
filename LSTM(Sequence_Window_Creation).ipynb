#Step 2: Create Dataset

```python
data=np.array([10,20,30,40,50,60])
print(data)
```

#Step 3: Decide Window Size

```python
window_size = 3
```

What does this mean?

The model will look at 3 previous values to predict the next value.

Visual

10 20 30 → ?


Answer

40

#Step 4: Create Empty Lists
X = []
y = []

These lists will store:

Input sequences
Output values

Initially

X = []

```python
X = []
y = []
```

#Step 5: Create Sequences

```python
for i in range(len(data) - window_size):

    X.append(data[i:i+window_size])

    y.append(data[i+window_size])
```
