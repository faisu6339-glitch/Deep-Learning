#Step 5: Create Sequences

````python
for i in range(len(data) - window_size):

    X.append(data[i:i+window_size])

    y.append(data[i+window_size])
````

### Step 1: Understanding `range(len(data) - window_size)`

First, let's calculate `len(data)`:

```
len(data)
```
Output:
```
6
```
because there are 6 numbers in our `data` array.

Now, `len(data) - window_size` becomes:

```
6 - 3 = 3
```
So, the loop effectively becomes:

```python
for i in range(3):
```
This means the variable `i` will take on the values `0`, `1`, and `2`. Only 3 iterations are needed.

---

### Iteration 1: `i = 0`

**First line: `X.append(data[i:i+window_size])`**

Substitute the values:

```python
X.append(data[0:3])
```
Remember Python slicing: `start : stop` includes the start index but excludes the stop index.

Our data and its indices are:

| Index | 0  | 1  | 2  | 3  | 4  | 5  |
|-------|----|----|----|----|----|----|
| Value | 10 | 20 | 30 | 40 | 50 | 60 |

So, `data[0:3]` takes the values at index 0, 1, and 2:

```
[10, 20, 30]
```
Now, `X.append([10,20,30])` results in:

```
X = [
    [10, 20, 30]
]
```

**Second line: `y.append(data[i+window_size])`**

Substitute values:

```python
y.append(data[0+3])  # which is data[3]
```
Index 3 contains the value `40`.

So, `y.append(40)` results in:

```
y = [
    40
]
```

**After Iteration 1:**

Input: `[10, 20, 30]`
Output: `40`

---

### Visual Explanation: The Sliding Window

Imagine a window of size `window_size` (which is 3) moving over your data:

**Window 1:**
`[10 20 30]` 40 50 60

*   **Input (X):** `[10, 20, 30]`
*   **Output (y):** `40`

**Window 2:**
10 `[20 30 40]` 50 60

*   **Input (X):** `[20, 30, 40]`
*   **Output (y):** `50`

**Window 3:**
10 20 `[30 40 50]` 60

*   **Input (X):** `[30, 40, 50]`
*   **Output (y):** `60`

---

### Why do we use `i + window_size` for `y`?

The goal of this data preparation is to create sequences where a model learns to predict the _next_ value in the sequence given the _previous_ `window_size` values. Therefore:

*   The first `window_size` values (`data[i:i+window_size]`) form the input sequence (`X`).
*   The value immediately following this sequence (`data[i+window_size]`) is the target output (`y`) that the model should predict.

This technique is commonly known as a **sliding window** or **lagged features** approach, and it's a standard way to prepare sequential data for time series forecasting or sequence prediction tasks with models like LSTMs.
