```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense

# -------------------------
# Dataset
# -------------------------

X = np.array([
    [1,2,3],
    [2,3,4],
    [3,4,5],
    [4,5,6]
])

y = np.array([4,5,6,7])

# RNN expects:
# (samples, timesteps, features)

X = X.reshape((4,3,1))

# -------------------------
# Build Model
# -------------------------

model = Sequential([
    SimpleRNN(32,
              activation='tanh',
              input_shape=(3,1)),
    Dense(1)
])

model.compile(
    optimizer='adam',
    loss='mse',
    metrics=['mae']
)

model.summary()

# -------------------------
# Train
# -------------------------

model.fit(
    X,
    y,
    epochs=200,
    verbose=1
)

# -------------------------
# Prediction
# -------------------------

test = np.array([[5,6,7]])
test = test.reshape((1,3,1))

prediction = model.predict(test)

print("Prediction:", prediction)
```

#Program 3: Predict Next Number

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN
from tensorflow.keras.layers import Dense

X = np.array([
    [1,2,3],
    [2,3,4],
    [3,4,5],
    [4,5,6]
])

y = np.array([
    4,
    5,
    6,
    7
])

X = X.reshape((4,3,1))

model = Sequential([
    SimpleRNN(
        32,
        activation='tanh',
        input_shape=(3,1)
    ),
    Dense(1)
])

model.compile(
    optimizer='adam',
    loss='mse'
)

model.fit(
    X,
    y,
    epochs=300
)

test = np.array([[5,6,7]])
test = test.reshape((1,3,1))

prediction = model.predict(test)

print(prediction)
```
