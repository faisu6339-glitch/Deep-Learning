## Backpropagation in Convolutional Neural Networks (CNNs)

Backpropagation is the fundamental algorithm for training artificial neural networks, including CNNs. It's an efficient way to calculate the gradient of the loss function with respect to the network's weights, allowing the network to adjust its parameters to minimize error.

### How Backpropagation Works in CNNs

In a standard feedforward neural network, backpropagation involves propagating the error backward through the layers, adjusting weights based on their contribution to the error. For CNNs, this process is similar but has specific considerations due to the unique operations (convolution, pooling) involved.

1.  **Forward Pass**: Input data passes through the convolutional layers, activation functions, pooling layers, and finally fully connected layers to produce an output prediction.

2.  **Loss Calculation**: The predicted output is compared to the actual target output, and a loss function (e.g., Cross-Entropy for classification, Mean Squared Error for regression) calculates the error.

3.  **Backward Pass (Gradient Calculation)**:
    *   **Output Layer**: The gradients of the loss with respect to the output layer's activations and weights are calculated first.
    *   **Fully Connected Layers**: Standard backpropagation applies to these layers, similar to an MLP.
    *   **Pooling Layers (e.g., Max Pooling)**: The gradient is passed *only* to the neuron that had the maximum activation during the forward pass. All other neurons in the receptive field receive a gradient of zero. This means the error signal is sparse.
    *   **Convolutional Layers**: This is where it gets interesting.
        *   **Weight Gradients**: To update the filter weights, the error signal from the subsequent layer is convolved with the input to the current convolutional layer. This is essentially a *"backward convolution"*.
        *   **Input Gradients (for previous layer)**: To pass the error backward to the previous layer, the filters themselves are *rotated 180 degrees* and then convolved with the error signal from the current layer. This effectively performs an upsampling and convolution operation to spread the error.

4.  **Parameter Update**: An optimization algorithm (e.g., Stochastic Gradient Descent, Adam, RMSprop) uses these calculated gradients to adjust the weights and biases of all layers, aiming to reduce the loss in subsequent iterations.

### Key Components and Their Role in Backpropagation

#### 1. Activation Functions
Activation functions (like ReLU, Sigmoid, Tanh) introduce non-linearity, allowing the network to learn complex patterns. During backpropagation, their derivatives are crucial for calculating how much each neuron's activation contributed to the overall error. For example, the derivative of ReLU is 1 for positive inputs and 0 for negative inputs, meaning only active neurons propagate gradients.

#### 2. Loss Functions
The choice of loss function directly impacts the error signal generated. Common loss functions include:
*   **Cross-Entropy Loss**: Widely used for classification tasks. It penalizes incorrect predictions more heavily.
*   **Mean Squared Error (MSE)**: Typically used for regression tasks. It measures the average squared difference between predicted and actual values.
The derivative of the loss function is the starting point for backpropagation, defining the initial error signal that propagates backward through the network.

### In Summary

Backpropagation in CNNs is a sophisticated application of the chain rule of calculus. It efficiently computes the gradients needed to update the network's convolutional filters, pooling operations, and fully connected layer weights, enabling CNNs to learn powerful feature representations from data for tasks like image classification, object detection, and more.

# Gradient Descent Formula

\[
W_{new} = W_{old} - \eta \frac{\partial L}{\partial W}
\]

Where:
- \(W\) = weight
- \(\eta\) = learning rate
- \(L\) = loss function
- \(\frac{\partial L}{\partial W}\) = gradient

# Cross Entropy Loss

\[
L = -\sum y \log(\hat{y})
\]

Where:
- \(y\) = actual label
- \(\hat{y}\) = predicted output

# Chain Rule in Backpropagation

\[
\frac{\partial L}{\partial W}
=
\frac{\partial L}{\partial O}
\times
\frac{\partial O}{\partial W}
\]

Where:
- \(L\) = Loss
- \(O\) = Output
- \(W\) = Weight

# ReLU Activation Function

\[
f(x) = \max(0, x)
\]

## ReLU Gradient

- If \(x > 0\), gradient = 1
- If \(x \leq 0\), gradient = 0

## Kernel Gradient Formula

\[
\frac{\partial L}{\partial K}
=
X * \frac{\partial L}{\partial Y}
\]

Where:
- \(K\) = kernel
- \(X\) = input image
- \(Y\) = output feature map

---

## Input Gradient Formula

\[
\frac{\partial L}{\partial X}
=
\frac{\partial L}{\partial Y} * K_{rotated}
\]

Kernel is rotated by 180° during backward convolution.

#Understanding Backpropagation in CNN with Code

````python
import torch
import torch.nn as nn
import torch.optim as optim
````

````python
# Input Image
x = torch.tensor([[[

    [1., 2., 3., 0., 1., 2.],
    [0., 1., 2., 3., 1., 0.],
    [1., 0., 1., 2., 3., 1.],
    [2., 1., 0., 1., 2., 3.],
    [3., 2., 1., 0., 1., 2.],
    [0., 1., 2., 1., 0., 1.]

]]])
print(x.shape)
````

#3. Create CNN Model

Now create:

Convolution
ReLU
Pooling
Fully Connected Layer

````python
class SimpleCNN(nn.Module):

    def __init__(self):

        super(SimpleCNN, self).__init__()

        # Convolution Layer
        self.conv = nn.Conv2d(
            in_channels=1,
            out_channels=1,
            kernel_size=3
        )

        # ReLU
        self.relu = nn.ReLU()

        # Max Pooling
        self.pool = nn.MaxPool2d(kernel_size=2)

        # Fully Connected Layer
        self.fc = nn.Linear(4, 1)

    def forward(self, x):

        print("\nINPUT IMAGE")
        print(x)

        # Convolution
        x = self.conv(x)

        print("\nAFTER CONVOLUTION")
        print(x)

        # ReLU
        x = self.relu(x)

        print("\nAFTER RELU")
        print(x)

        # Pooling
        x = self.pool(x)

        print("\nAFTER POOLING")
        print(x)

        # Flatten
        x = x.view(x.size(0), -1)

        print("\nAFTER FLATTEN")
        print(x)

        # Fully Connected
        x = self.fc(x)

        print("\nFINAL OUTPUT")
        print(x)

        return x
````

#4. Create Model

````python
model = SimpleCNN()
````

#5. Define Loss Function

We use:

Binary Cross Entropy

````python
criterion = nn.BCEWithLogitsLoss()
````

#6. Define Optimizer

Optimizer updates weights.

````python
optimizer = optim.SGD(model.parameters(), lr=0.01)
````

#7. Create Target Label

````python
target = torch.tensor([[1.]])
````

#8. Forward Pass

````python
output = model(x)
````

#9. Compute Loss

````python
loss = criterion(output, target)

print("\nLOSS")
print(loss)
````

#10. Backpropagation Begins

````python
optimizer.zero_grad()

loss.backward()
````

#Convolution Filter Gradients

````python
print("\nCONVOLUTION FILTER GRADIENTS")

print(model.conv.weight.grad)
````

#Fully Connected Layer Gradients

````python
print("\nFC LAYER GRADIENTS")

print(model.fc.weight.grad)
````

#12. Update Weights

````python
optimizer.step()
````

#13. Print Updated Weights

````python
print("\nUPDATED CONVOLUTION FILTER")

print(model.conv.weight)
````
