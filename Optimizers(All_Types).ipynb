### What are Optimizers?

In the context of deep learning, an **optimizer** is an algorithm or function that adjusts the parameters of the model (primarily weights and biases) in order to minimize the error (loss function) and improve the model's performance. The goal is to find the optimal set of parameters that make the model's predictions as accurate as possible.

### The Core Idea: Gradient Descent

Most optimizers are built upon the principle of **Gradient Descent**. Imagine the loss function as a mountainous landscape, and you want to find the lowest point (the minimum loss). Gradient descent is like taking steps down the steepest slope. The 'gradient' tells you the direction of the steepest ascent, so you move in the opposite direction.

The update rule for a parameter `θ` (theta) in basic gradient descent is:

`θ = θ - learning_rate * ∇J(θ)`

Where:
*   `θ`: The parameter being updated (e.g., a weight or bias).
*   `learning_rate (η)`: A hyperparameter that determines the size of the steps taken down the gradient. A small learning rate leads to slow convergence, while a large one might overshoot the minimum.
*   `∇J(θ)`: The gradient of the loss function `J` with respect to the parameter `θ`. It indicates the direction and magnitude of the steepest ascent.

### Types of Gradient Descent:

1.  **Batch Gradient Descent (BGD)**:
    *   Calculates the gradient of the loss function with respect to the entire training dataset for each update.
    *   **Pros**: Guaranteed to converge to the global minimum for convex error surfaces and to a local minimum for non-convex surfaces. Updates are stable.
    *   **Cons**: Can be very slow and computationally expensive for large datasets because it processes all training examples before making an update.

2.  **Stochastic Gradient Descent (SGD)**:
    *   Calculates the gradient and updates the parameters for *each individual training example*.
    *   **Pros**: Much faster than BGD for large datasets. Its frequent updates allow for rapid changes and can help escape shallow local minima (due to noise in updates).
    *   **Cons**: High variance in updates, which can cause the loss function to fluctuate and never truly converge to a minimum, instead oscillating around it.

3.  **Mini-Batch Gradient Descent (MBGD)**:
    *   The most common and practical approach. It processes data in small batches (e.g., 32, 64, 128 examples) to calculate the gradient and update parameters.
    *   **Pros**: Balances the stability of BGD with the speed of SGD. Reduces variance in updates compared to SGD, leading to more stable convergence. Exploits GPU parallelism efficiently.
    *   **Cons**: Requires tuning the batch size.

### Advanced Optimizers (Adaptive Learning Rate Methods):

These optimizers address the limitations of basic SGD (e.g., choosing a suitable learning rate, slow convergence, getting stuck in saddle points or local minima).

1.  **SGD with Momentum**:
    *   Introduces a 'momentum' term, which accumulates an exponentially decaying moving average of past gradients. This helps accelerate SGD in the relevant direction and dampens oscillations.
    *   **Idea**: If the gradient consistently points in the same direction, momentum builds up, and the optimizer takes larger steps. If gradients change direction frequently, momentum helps smooth out the steps.
    *   **Update Rule**: `v_t = γ * v_{t-1} + η * ∇J(θ)`
                    `θ = θ - v_t`
        Where `v_t` is the velocity, and `γ` is the momentum term (typically 0.9).

2.  **Nesterov Accelerated Gradient (NAG)**:
    *   A variation of momentum that looks ahead. Instead of calculating the gradient at the current position, it calculates it at an approximated future position (where the momentum would take it).
    *   **Idea**: This 'look-ahead' helps to correct the path before taking the actual step, often leading to faster and more stable convergence than standard momentum.
    *   **Update Rule**: `v_t = γ * v_{t-1} + η * ∇J(θ - γ * v_{t-1})`
                    `θ = θ - v_t`

3.  **Adagrad (Adaptive Gradient)**:
    *   Adapts the learning rate for each parameter individually. It scales the learning rate by the inverse of the square root of the sum of squared past gradients.
    *   **Idea**: Parameters that have infrequent updates (e.g., sparse features) get larger learning rates, while parameters with frequent updates get smaller learning rates.
    *   **Pros**: Good for sparse data. No need to manually tune the learning rate as much.
    *   **Cons**: The learning rate can become very small and eventually close to zero after many updates, effectively stopping further learning.

4.  **Adadelta**:
    *   An extension of Adagrad that aims to overcome its aggressive, monotonically decreasing learning rate. It restricts the window of accumulated past gradients to a fixed size.
    *   **Idea**: Instead of accumulating all past squared gradients, Adadelta accumulates a decaying average of past squared gradients. It also removes the need for setting an initial learning rate.
    *   **Pros**: Doesn't require setting a default learning rate. Addresses Adagrad's vanishing learning rate issue.

5.  **RMSprop (Root Mean Square Propagation)**:
    *   Developed by Geoff Hinton. Similar to Adadelta, it addresses Adagrad's diminishing learning rates by using a moving average of squared gradients.
    *   **Idea**: It divides the learning rate by the square root of the exponential moving average of squared gradients. This helps to normalize the step size based on the magnitude of recent gradients.
    *   **Pros**: Effective in non-stationary settings (e.g., recurrent neural networks).
    *   **Cons**: Still requires an initial learning rate.

6.  **Adam (Adaptive Moment Estimation)**:
    *   Combines the best of RMSprop and Momentum. It computes adaptive learning rates for each parameter.
    *   **Idea**: Adam keeps an exponentially decaying average of past gradients (like momentum) AND an exponentially decaying average of past squared gradients (like RMSprop/Adagrad). It also includes bias-correction terms for these moving averages.
    *   **Pros**: Widely considered one of the most robust and efficient optimizers. Generally performs well across a wide range of tasks and architectures. Less sensitive to hyperparameter tuning.
    *   **Cons**: Can sometimes generalize slightly worse than SGD with momentum in specific scenarios, though this is often negligible.

7.  **AdamW**:
    *   A variant of Adam that decouples weight decay from the gradient update. In standard Adam, weight decay (L2 regularization) is applied directly to the gradients, which isn't always optimal.
    *   **Idea**: AdamW separates the weight decay term from the adaptive gradient scaling. This often leads to better generalization performance, especially in models with strong regularization.
    *   **Pros**: Often improves generalization over standard Adam, especially in tasks like computer vision.

### Learning Rate Scheduling/Decay

Even with adaptive optimizers, varying the learning rate during training can be beneficial. **Learning rate scheduling** refers to strategies that change the learning rate over time. Common schedules include:

*   **Step decay**: Reduce the learning rate by a factor every few epochs.
*   **Exponential decay**: Reduce the learning rate exponentially over time.
*   **Cosine annealing**: The learning rate follows a cosine curve, starting high, decreasing, and potentially increasing again.
*   **Warm-up**: Start with a very small learning rate and gradually increase it before applying a decay schedule.

### Choosing an Optimizer

*   **Start with Adam**: For most deep learning tasks, Adam (or AdamW) is a strong default choice due to its robustness and good performance across various scenarios.
*   **SGD with Momentum/Nesterov**: If you have time for extensive hyperparameter tuning or observe that Adam is overfitting, SGD with momentum or NAG can sometimes yield slightly better generalization, especially in image classification tasks.
*   **Consider Data Sparsity**: Adagrad or RMSprop might be considered for very sparse data, though Adam usually handles this well too.
*   **Experiment**: The best optimizer can be task-dependent. It's always a good practice to experiment with a few different optimizers and learning rate schedules.

Optimizers are a fundamental component of training deep learning models, enabling them to navigate complex loss landscapes and find effective solutions. Understanding their mechanisms helps in debugging training issues and improving model performance.

### Adam (Adaptive Moment Estimation)

Adam is one of the most widely used and highly effective optimizers in deep learning. It cleverly combines the best aspects of two other adaptive learning rate methods: RMSprop and Momentum.

**Core Idea:**

Adam computes adaptive learning rates for each parameter individually by keeping an exponentially decaying average of past gradients (like Momentum) AND an exponentially decaying average of past squared gradients (like RMSprop/Adagrad). It also includes bias-correction terms for these moving averages, which is particularly important during the initial steps of training.

**How it works (Intuition):**

1.  **Momentum-like part (First Moment Estimation):** Adam keeps track of an exponential moving average of the gradients, which effectively acts like a 'velocity' pushing the parameters in a consistent direction. This helps accelerate convergence and dampens oscillations, similar to how momentum works.

    *   Mathematically, it calculates `m_t` (the first moment vector) as:
        `m_t = β1 * m_{t-1} + (1 - β1) * g_t`
        where `g_t` is the gradient at time `t`, and `β1` is the exponential decay rate for the first moment estimate (typically 0.9).

2.  **RMSprop-like part (Second Moment Estimation):** Adam also maintains an exponential moving average of the *squared* gradients. This provides adaptive learning rates, meaning parameters with large gradients (e.g., in sparse data) get smaller updates, while parameters with small gradients get larger updates. This helps manage the step size in different directions, similar to RMSprop.

    *   Mathematically, it calculates `v_t` (the second moment vector) as:
        `v_t = β2 * v_{t-1} + (1 - β2) * g_t^2`
        where `g_t^2` is the element-wise square of the gradient, and `β2` is the exponential decay rate for the second moment estimate (typically 0.999).

3.  **Bias Correction:** Because `m_t` and `v_t` are initialized as zeros, they are biased towards zero, especially during the initial time steps. Adam includes bias-correction terms to counteract this, providing more accurate estimates of the first and second moments early in training.

    *   Bias-corrected first moment estimate:
        `m_hat_t = m_t / (1 - β1^t)`

    *   Bias-corrected second moment estimate:
        `v_hat_t = v_t / (1 - β2^t)`

4.  **Parameter Update:** Finally, the parameters are updated using the bias-corrected first and second moment estimates:

    *   `θ_t = θ_{t-1} - η * m_hat_t / (sqrt(v_hat_t) + ε)`
        Where:
        *   `θ_t`: The parameter at time `t`.
        *   `η` (learning rate): The overall step size.
        *   `ε`: A small constant (e.g., 1e-8) added for numerical stability to prevent division by zero.

**Pros of Adam:**

*   **Robustness:** Widely considered one of the most robust and efficient optimizers. It generally performs well across a wide range of tasks and architectures (e.g., CNNs, RNNs, Transformers).
*   **Efficiency:** Computationally efficient, requiring relatively little memory.
*   **Less Sensitive to Hyperparameter Tuning:** Compared to SGD, Adam often requires less manual tuning of the learning rate and other hyperparameters, making it a good default choice.
*   **Adaptive Learning Rates:** Automatically adjusts learning rates for different parameters, which is beneficial for sparse gradients or non-stationary objectives.

**Cons of Adam:**

*   **Generalization Gap (sometimes):** In some specific scenarios, particularly in image classification, SGD with momentum (when carefully tuned) has been observed to achieve slightly better generalization performance than Adam. However, this gap is often negligible, and Adam still provides a faster and more stable convergence initially.
*   **Potential for Suboptimal Solutions (rare):** There have been theoretical and empirical observations where Adam can sometimes converge to suboptimal solutions, especially if the `β2` term is not chosen carefully, leading to overly aggressive adaptive learning rates. This issue is less common with newer variants like AdamW.

**Key Hyperparameters:**

*   **`η` (learning_rate):** The base learning rate. Common values are 0.001 or 0.0001.
*   **`β1`:** The exponential decay rate for the first moment estimates. Typically 0.9.
*   **`β2`:** The exponential decay rate for the second moment estimates. Typically 0.999.
*   **`ε` (epsilon):** A small constant for numerical stability. Typically 1e-8.

In summary, Adam is a powerful and popular optimizer due to its adaptive nature and combination of momentum and RMSprop principles. It's often the first optimizer to try for most deep learning problems.

### Why we use Optimizers and what they actually do:

Optimizers are crucial in deep learning because they dictate **how** a neural network learns. They actually do the heavy lifting of adjusting the model's internal parameters (weights and biases) during the training phase.

Here's a breakdown of what they actually do and why we use them:

### Why we use Optimizers:

1.  **Minimize Loss (Error):** The primary goal of training a neural network is to make its predictions as accurate as possible. This is quantified by a 'loss function' (or cost function), which measures how far off the model's predictions are from the actual true values. Optimizers' main job is to find the set of weights and biases that minimize this loss function.

2.  **Efficient Learning:** Without optimizers, a neural network would have no systematic way to improve. They provide the mechanism to iteratively refine the model's parameters based on the feedback from the loss function, guiding the model towards better performance.

3.  **Navigate Complex Landscapes:** The 'loss landscape' (a plot of loss value against parameter values) can be very complex, high-dimensional, and non-convex, with many peaks, valleys, and saddle points. Optimizers help the model navigate this landscape to find a 'good enough' minimum, rather than getting stuck in local minima or wandering aimlessly.

### What they actually do:

Optimizers use algorithms, mostly based on **gradient descent**, to perform the following steps repeatedly during training:

1.  **Calculate the Gradient:** For a given set of input data, the model makes a prediction, and the loss function calculates the error. Then, the optimizer calculates the 'gradient' of the loss function with respect to each parameter (weight and bias) in the network. The gradient essentially tells us two things:
    *   **Direction:** In which direction should we change a parameter to *increase* the loss most rapidly?
    *   **Magnitude:** How sensitive is the loss to changes in that specific parameter?

2.  **Update Parameters:** Based on these calculated gradients, the optimizer adjusts each parameter in the opposite direction of the gradient (to *decrease* the loss). The size of this adjustment is controlled by the 'learning rate'.

    The general update rule is:
    `new_parameter = old_parameter - (learning_rate * gradient)`

3.  **Iterate:** This process of calculating gradients and updating parameters is repeated many times (over many batches of data and many epochs) until the model's loss is sufficiently low, or its performance on a validation set stops improving.

Different optimizers introduce variations to this basic gradient descent process (e.g., adding momentum, adapting learning rates per parameter, using moving averages of gradients) to achieve faster convergence, more stable training, or better generalization.

### Implementing Optimizers in PyTorch

In PyTorch, optimizers are part of the `torch.optim` module. You typically instantiate an optimizer by passing it the model's parameters and a learning rate. Then, in your training loop, you'll perform these steps:

1.  **Zero the gradients:** `optimizer.zero_grad()`
2.  **Compute loss:** `loss.backward()`
3.  **Update parameters:** `optimizer.step()`

Let's see an example with a simple linear regression model.

```python
import torch
import torch.nn as nn
import torch.optim as optim

# 1. Define a simple model
class LinearRegression(nn.Module):
    def __init__(self):
        super(LinearRegression, self).__init__()
        self.linear = nn.Linear(1, 1) # One input feature, one output

    def forward(self, x):
        return self.linear(x)

# 2. Generate some dummy data
X = torch.randn(100, 1) * 10 # 100 samples, 1 feature
y = 2 * X + 3 + torch.randn(100, 1) # y = 2x + 3 + noise

# 3. Instantiate the model, loss function, and optimizer
model = LinearRegression()
criterion = nn.MSELoss() # Mean Squared Error Loss

# --- Example 1: SGD Optimizer ---
print("\n--- Training with SGD Optimizer ---")
optimizer_sgd = optim.SGD(model.parameters(), lr=0.01)

num_epochs = 100
for epoch in range(num_epochs):
    # Forward pass
    outputs = model(X)
    loss = criterion(outputs, y)

    # Backward and optimize
    optimizer_sgd.zero_grad() # Zero the gradients
    loss.backward()         # Compute gradients
    optimizer_sgd.step()    # Update weights

    if (epoch+1) % 20 == 0:
        print(f'Epoch [{epoch+1}/{num_epochs}], Loss: {loss.item():.4f}')

print(f"Final SGD Model Parameters: {list(model.parameters())}")

# Reset model for next optimizer demo
model = LinearRegression()

# --- Example 2: Adam Optimizer ---
print("\n--- Training with Adam Optimizer ---")
optimizer_adam = optim.Adam(model.parameters(), lr=0.01)

for epoch in range(num_epochs):
    # Forward pass
    outputs = model(X)
    loss = criterion(outputs, y)

    # Backward and optimize
    optimizer_adam.zero_grad()
    loss.backward()
    optimizer_adam.step()

    if (epoch+1) % 20 == 0:
        print(f'Epoch [{epoch+1}/{num_epochs}], Loss: {loss.item():.4f}')

print(f"Final Adam Model Parameters: {list(model.parameters())}")
```

### Implementing Optimizers in TensorFlow

In TensorFlow, optimizers are available under `tf.keras.optimizers`. When compiling a Keras model, you specify the optimizer directly. For custom training loops, you apply gradients using `optimizer.apply_gradients()`.

Let's see an example with a simple linear regression model using Keras's sequential API.

```python
import tensorflow as tf
import numpy as np

# 1. Define a simple model
model_tf = tf.keras.Sequential([
    tf.keras.layers.Dense(units=1, input_shape=[1]) # One input, one output
])

# 2. Generate some dummy data
X_tf = np.random.rand(100, 1) * 10
y_tf = 2 * X_tf + 3 + np.random.randn(100, 1) * 0.5

# --- Example 1: SGD Optimizer ---
print("\n--- Training with SGD Optimizer (TensorFlow) ---")
# Compile the model with SGD
model_tf.compile(optimizer=tf.keras.optimizers.SGD(learning_rate=0.01), loss='mse')

# Train the model
history_sgd = model_tf.fit(X_tf, y_tf, epochs=100, verbose=0)
print(f"Final SGD Model Loss: {history_sgd.history['loss'][-1]:.4f}")
print(f"Final SGD Model Weights: {model_tf.get_weights()}")

# Reset model for next optimizer demo
model_tf = tf.keras.Sequential([
    tf.keras.layers.Dense(units=1, input_shape=[1])
])

# --- Example 2: Adam Optimizer (TensorFlow) ---
print("\n--- Training with Adam Optimizer (TensorFlow) ---")
# Compile the model with Adam
model_tf.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.01), loss='mse')

# Train the model
history_adam = model_tf.fit(X_tf, y_tf, epochs=100, verbose=0)
print(f"Final Adam Model Loss: {history_adam.history['loss'][-1]:.4f}")
print(f"Final Adam Model Weights: {model_tf.get_weights()}")
```

#Program (Cats VS Dogs)

```python
# ==========================================
# IMPORT LIBRARIES
# ==========================================

import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.layers import (
    Dense,
    Dropout,
    GlobalAveragePooling2D,
    BatchNormalization
)
from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import (
    EarlyStopping,
    ModelCheckpoint,
    ReduceLROnPlateau
)
import matplotlib.pyplot as plt

# ==========================================
# IMAGE SETTINGS
# ==========================================

IMG_SIZE = (224, 224)
BATCH_SIZE = 32

# ==========================================
# DATA AUGMENTATION
# ==========================================

train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    zoom_range=0.2,
    shear_range=0.2,
    horizontal_flip=True
)

val_test_datagen = ImageDataGenerator(
    rescale=1./255
)

# ==========================================
# LOAD DATASET
# ==========================================

train_data = train_datagen.flow_from_directory(
    "dataset/train",
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='binary'
)

val_data = val_test_datagen.flow_from_directory(
    "dataset/validation",
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='binary'
)

test_data = val_test_datagen.flow_from_directory(
    "dataset/test",
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='binary',
    shuffle=False
)

# ==========================================
# LOAD PRETRAINED MODEL
# ==========================================

base_model = MobileNetV2(
    weights='imagenet',
    include_top=False,
    input_shape=(224, 224, 3)
)

# ==========================================
# FREEZE BASE MODEL
# ==========================================

for layer in base_model.layers:
    layer.trainable = False

# ==========================================
# BUILD CUSTOM MODEL
# ==========================================

x = base_model.output

x = GlobalAveragePooling2D()(x)

x = BatchNormalization()(x)

x = Dense(128, activation='relu')(x)

x = Dropout(0.5)(x)

output = Dense(1, activation='sigmoid')(x)

model = Model(inputs=base_model.input, outputs=output)

# ==========================================
# COMPILE MODEL USING ADAM
# ==========================================

model.compile(
    optimizer=Adam(
        learning_rate=0.001
    ),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# ==========================================
# CALLBACKS
# ==========================================

early_stop = EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True
)

checkpoint = ModelCheckpoint(
    "best_model.h5",
    monitor='val_accuracy',
    save_best_only=True
)

reduce_lr = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.2,
    patience=2,
    min_lr=0.00001
)

# ==========================================
# TRAIN MODEL
# ==========================================

history = model.fit(
    train_data,
    validation_data=val_data,
    epochs=10,
    callbacks=[
        early_stop,
        checkpoint,
        reduce_lr
    ]
)

# ==========================================
# EVALUATE MODEL
# ==========================================

test_loss, test_accuracy = model.evaluate(test_data)

print("\nTest Accuracy:", test_accuracy)

# ==========================================
# SAVE MODEL
# ==========================================

model.save("cats_vs_dogs_model.h5")

print("\nModel Saved Successfully!")

# ==========================================
# PLOT ACCURACY GRAPH
# ==========================================

plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])

plt.title("Model Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")

plt.legend([
    "Training Accuracy",
    "Validation Accuracy"
])

plt.show()

# ==========================================
# PLOT LOSS GRAPH
# ==========================================

plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])

plt.title("Model Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.legend([
    "Training Loss",
    "Validation Loss"
])

plt.show()
```
