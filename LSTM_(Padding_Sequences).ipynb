### What is Padding and Why Do We Need It?

Padding is a technique used in sequence processing, especially in deep learning and natural language processing, to make all input sequences the same length. Here's why it's needed:

*   **Consistent Input Size:** Many machine learning models, particularly neural networks, require fixed-size input arrays or tensors. Since natural sequences (like sentences or time series) often vary in length, padding ensures that every sequence conforms to the required input dimension.
*   **Batch Processing:** When you process data in batches (groups of sequences), all sequences within a batch must have the same length to be efficiently processed by hardware like GPUs.
*   **Mathematical Operations:** Padding allows for vectorized operations across entire batches without complex dynamic memory allocation or control flow for varying sequence lengths.

There are two main types of padding:

*   **Pre-padding:** Adding zeros (or another designated padding value) to the beginning of the sequence.
*   **Post-padding:** Adding zeros to the end of the sequence.

**Truncating** is the opposite of padding, where longer sequences are cut down to a maximum desired length, also to ensure uniform input size.

#Suppose we have three sentences.

I love AI

Deep Learning is amazing

Python

After tokenization:

I love AI
↓

[2, 5, 8]

Deep Learning is amazing
↓

[3, 6, 9, 7]

Python
↓

[10]

Notice the lengths:
Sentence 1 → Length = 3

Sentence 2 → Length = 4

Sentence 3 → Length = 1

An LSTM cannot process batches where each sequence has a different length.

So we use padding.

````python
from tensorflow.keras.preprocessing.sequence import pad_sequences
````

````python
sequences=[
    [2,5],
    [ 3,6,9,7],
    [10,20,30,40,50],
    [10]
]
print(sequences)
````

#Example 1: Default Padding (Pre Padding)

````python
padded = pad_sequences(sequences)

print(padded)
````

#Example 2: Post Padding

Instead of adding zeros at the beginning, we add them at the end.

````python
padded = pad_sequences(
    sequences,
    padding='post'
)

print(padded)
````

#Example 3: Specify Maximum Length

````python
padded = pad_sequences(
    sequences,
    maxlen=6
)

print(padded)
````

#Example 4: Post Padding + Maximum Length

````python
padded = pad_sequences(
    sequences,
    maxlen=6,
    padding='post'
)

print(padded)
````

#What is Truncating?

Suppose the sequence is too long.

[2,5,8,9,4,6,1]

But the model only accepts 5 words.

The extra words must be removed.

This process is called truncating.

#Example 5: Default Truncating(Pre)

````python
sequence = [
    [2,5,8,9,4,6,1]
]

padded = pad_sequences(
    sequence,
    maxlen=5
)

print(padded)
````

#Example 6: Post Truncating

````python
sequence = [
    [2,5,8,9,4,6,1]
]

padded = pad_sequences(
    sequence,
    maxlen=5,
    truncating='post'
)

print(padded)
````

#Combine Padding and Truncating

````python
sequences = [
    [2,5],
    [3,6,9,7],
    [8,9,4,2,1,5,7]
]

padded = pad_sequences(
    sequences,
    maxlen=5,
    padding='post',
    truncating='post'
)

print(padded)
````
